```python
import sqlite3
import pandas as pd
```

```python
def query_db(db_path, query):
    with sqlite3.connect(db_path) as conn:
        result = pd.read_sql_query(query, conn)
    return result
```

```python
db_path = "traces.db"
```

```python
query = "SELECT * FROM spans"
result = query_db(db_path, query)
print(result)
```

```python
# Unique span names
query = "SELECT DISTINCT name FROM spans"
result = query_db(db_path, query)
print(result)
```

```python
# Spans that take longer ecluding the rag span
query = """
SELECT name, start_time, end_time, (end_time - start_time) / 1000000.0 AS duration_ms
FROM spans
WHERE name != 'rag'
ORDER BY end_time - start_time DESC
LIMIT 5
"""
result = query_db(db_path, query)
print(result)
```

```python
query = """
SELECT input_tokens
FROM spans
WHERE name = 'llm'
ORDER BY rowid DESC
LIMIT 4;
"""
result = query_db(db_path, query)
print(result)
```
